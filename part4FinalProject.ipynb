Using error propagation, find out what the root mean squared uncertainty in your measurement of D and the transit time is. Don't forget to take into account both the uncertainty in the data and the uncertainty from your fits.

```python
#Importing numpy | pulling QTable from astropy's table library | bring in the pLanet lightcurve file
import numpy as np
from astropy.table import QTable
Data = QTable.read("Planet_Lightcurve.fits", format='fits') 

#Uploading Data from Part-3
time_clean = Data['time [days]']
flux_clean = Data['flux']
transit_mask = (time_clean >= -0.04) & (time_clean <= 0.04)

#measuring the time gap between data points
time_gap = time_clean[1] - time_clean[0] 
time_uncertainty = np.sqrt(time_gap**2 + time_gap**2) #The final time uncertanity check
#tested the original points for error propagation

#Calculating D and its uncertainty using standard error propagation
depth = np.mean(flux_clean[~transit_mask]) - np.mean(flux_clean[transit_mask])
sigma_d = np.sqrt(np.std(flux_clean[~transit_mask])**2 + np.std(flux_clean[transit_mask])**2)

#Calculating transit and its uncertainty based of time range of the graph
transit_time = 0.04 - (-0.04)
transit_time_error = np.sqrt(time_gap**2 + time_gap**2)

#final values required for standard error propagation
print("_________________________________________________________")
print("transit time:", transit_time)
print("measurement of D:", depth)
print("combined uncertainty for transit:", time_uncertainty)
print("_________________________________________________________")
print("total combined uncertainty for transit:", transit_time_error)
print("total standard deviation for D:", sigma_d)
print("_________________________________________________________")
```
